When NumPy encounters 0 in np.log():

- It raises a RuntimeWarning: divide by zero encountered in log.
- It outputs -inf for that position in the array.

Why this is a problem in Machine Learning?

- If your target variable or feature contains 0, computing loss functions (like RMSE) or multiplying weights against -inf breaks calculations immediately (yielding NaN).


The Solution: 

np.log1p() instead of np.log(x), which computes $\ln(1 + x)$

```python
np.log1p([0, 1, 10, 1000, 100000])
```

```python
price_logs = np.log1p(df.msrp)
```

## Setting Validation Framework

```python
n = len(df)

n_val = int(n * 0.2)
n_test = int(n * 0.2)
n_train = n - n_val - n_test # better than int(n * 0.6)
```

```python
n, n_train, n_val, n_test
```

```python
df_train = df.iloc[n_train:]
df_val = df.iloc[n_train:n_train+n_val]
df_test = df.iloc[n_train+n_val:]
```

```python
df_train
```

## shuffle order

```python
idx = np.arange(n)
```

```python
np.random.seed(2) #for reproducibility
np.random.shuffle(idx)
```

```python
df_train = df.iloc[idx[:n_train]]
df_val = df.iloc[idx[n_train:n_train+n_val]]
df_test = df.iloc[idx[n_train+n_val:]]
```

```python
df_train
```

### reset index

```python
df_train = df_train.reset_index(drop=True)
df_val = df_val.reset_index(drop=True)
df_test = df_test.reset_index(drop=True)
```
