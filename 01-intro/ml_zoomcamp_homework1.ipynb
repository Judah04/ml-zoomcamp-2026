```python
df.head()
```

```python
df_asia = df[df['origin'] == 'Asia']
df_asia = df_asia[["vehicle_weight", "model_year"]]
df_asia.head(7)
```

```python
X = df_asia.head(7).values
X
```

```python
XTX = X.T.dot(X)
XTX
```

```python
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = XTX_inv.dot(X.T).dot(y)
w
```

```python
result_sum = w.sum()
result_sum
```
